## <font color="Red">1.3 Calculate gray image (Y) from color image</font>
<img src="IP01_1.png" width="600px"> 

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

imgBGR = cv2.imread("Commons.jpeg", cv2.IMREAD_COLOR)
img = imgBGR[:,:,::-1]

# Image size and informstion
nr, nc = img.shape[:2]
gray = np.zeros((nr,nc),dtype=np.uint8)

#-------------- Code Here ------------------#

# RGB to grayscale
gray = np.round(
    0.299 * img[:, :, 0] +
    0.587 * img[:, :, 1] +
    0.114 * img[:, :, 2]
).astype(np.uint8)

#-------------- Stop Here ------------------#

plt.figure(figsize=(12, 10))
plt.subplot(1,2,1)   
plt.title('Image color') 
plt.imshow(img)
plt.subplot(1,2,2)     
plt.title('Image gray') 
plt.imshow(gray, cmap=plt.cm.gray)        
```

## <font color="Red">1.4 Use for-loop to calculate Cb and Cr</font>

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

imgBGR = cv2.imread("Commons.jpeg", cv2.IMREAD_COLOR)
img = imgBGR[:,:,::-1]

# Image size and informstion
nr, nc = img.shape[:2]
Cb = np.zeros((nr,nc),dtype=np.uint8)
Cr = np.zeros((nr,nc),dtype=np.uint8)
#-------------- Code Here ------------------#
for i in range(nr):
    for j in range(nc):
        R = img[i, j, 0]
        G = img[i, j, 1]
        B = img[i, j, 2]
        Y = 0.299 * R + 0.587 * G + 0.114 * B
        Cb[i, j] = np.uint8(-0.1687 * R - 0.3313 * G + 0.5 * B + 128)
        Cr[i, j] = np.uint8(0.5 * R - 0.4187 * G - 0.0813 * B + 128)

#-------------- Stop Here ------------------#
plt.figure(figsize=(12, 10))
plt.subplot(1,3,1)   
plt.title('Image color') 
plt.imshow(img)
plt.subplot(1,3,2)     
plt.title('Image Cb') 
plt.imshow(Cb) 
plt.subplot(1,3,3)     
plt.title('Image Cr') 
plt.imshow(Cr)        
```
